```python
import pandas as pd
import matplotlib.pyplot as plt

BASE = "https://api.openf1.org/v1"
SESSION = 11420  # 2026 Las Vegas GP race

def get(endpoint, **params):
    query = "&".join(f"{k}={v}" for k, v in params.items())
    return pd.read_json(f"{BASE}/{endpoint}?{query}")

get("sessions", session_key=SESSION)
```

```python
drivers = get("drivers", session_key=SESSION)
names = dict(zip(drivers.driver_number, drivers.name_acronym))
colors = dict(zip(drivers.driver_number, "#" + drivers.team_colour.fillna("888888")))
drivers[["driver_number", "name_acronym", "team_name"]]
```

```python
laps = get("laps", session_key=SESSION)
laps["driver"] = laps.driver_number.map(names)

# drop safety-car and pit laps (slower than 107% of median)
clean = laps[laps.lap_duration < laps.lap_duration.median() * 1.07]

fig, ax = plt.subplots(figsize=(10, 5))
for num in [1, 4, 16, 63]:  # pick any driver numbers from Cell 2
    d = clean[clean.driver_number == num]
    ax.plot(d.lap_number, d.lap_duration, ".-", label=names.get(num), color=colors.get(num))
ax.set_xlabel("Lap"); ax.set_ylabel("Lap time (s)"); ax.legend()
plt.show()
```
